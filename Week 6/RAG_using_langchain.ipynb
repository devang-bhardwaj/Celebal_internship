# Retrieval-Augmented Generation (RAG) using LangChain

This notebook demonstrates the implementation of a RAG system using LangChain framework to create an intelligent Q&A system with the following components:

1. Document Loading and Processing
2. Vector Store Creation using FAISS
3. LLM Integration for Question Answering
4. RAG Pipeline Implementation

```python
# Import required libraries
from langchain.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS
from langchain.chains import RetrievalQA
from langchain.llms import OpenAI
```

## 1. Document Loading and Processing

We'll start by loading our documents and splitting them into manageable chunks.

```python
# Load PDF document
loader = PyPDFLoader("sample_document.pdf")
documents = loader.load()

# Split text into chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
texts = text_splitter.split_documents(documents)
```

## 2. Creating Vector Store

Now we'll create embeddings and store them in a FAISS vector store for efficient retrieval.

```python
# Initialize embeddings
embeddings = HuggingFaceEmbeddings()

# Create vector store
vectorstore = FAISS.from_documents(texts, embeddings)

# Save vector store
vectorstore.save_local("faiss_index")
```

## 3. Setting up the RAG Pipeline

Here we'll create a question-answering chain that combines the retrieval system with an LLM.

```python
# Initialize LLM
llm = OpenAI()

# Create QA chain
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=vectorstore.as_retriever()
)
```

## 4. Testing the RAG System

Let's test our RAG system with some sample questions.

```python
# Test questions
questions = [
    "What are the main topics covered in the document?",
    "Can you summarize the key findings?",
    "What recommendations are provided?"
]

# Get answers
for question in questions:
    print(f"Question: {question}")
    result = qa_chain.run(question)
    print(f"Answer: {result}\n")
```
